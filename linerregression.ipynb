```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
```

```python
df = pd.DataFrame({"studyhours" :[1,2,3,4,5,6,7,8,9],
                   "marks":[23,4,56,45,34,65,32,46,34]})

print(df)
```

```python
x = df[["studyhours"]]
y = df["marks"]

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
model = LinearRegression()
model.fit(x_train,y_train)
print(x_train)
print(y_train)
print(x_test)
print(y_test)
m = model.coef_[0]
b = model.intercept_
print("m value=",m)
print("b value = ",b)
print(f"equation:y={m:2f}x + {b:2f}")
```

```python
y_pre = model.predict(x_test)
comparison = pd.DataFrame({
    "study_hours":x_test["studyhours"].values,
    "marks":y_test.values,
    "predicated_marks":y_pre})
print(comparison)
```

```python
mse = mean_squared_error(y_test,y_pre)
print("mse=",mse)
rmse = np.sqrt(mse)
print("rmse=",rmse)
r2 = r2_score(y_test,y_pre)
print("re score=",r2)
```
