```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split

from sklearn.linear_model import LogisticRegression

from sklearn.preprocessing import StandardScaler

from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score,classification_report
```

```python
data = {
    "studyhours" :[1,2,3,4,5,6,7,8,9],
    "attendance" : [12,34,44,34,76,34,56,34,78],
    "internet" :["yes","no","no","yes","yes","no","no","yes","no"],
    "result" :["fail","pass","fail","fail","fail","pass","fail","fail","pass"]
}
df = pd.DataFrame(data)
print(df)
```

```python
print(df.head())
print(df.tail())
print(df.shape)
print(df.info())
print(df.describe())
```

```python
df["internet"] = df["internet"].map({
    "no" : 0,
    "yes" : 1})
print(df)
```

```python
print(df["internet"].unique())
```

```python
df["result"] = df["result"].map({
    "pass" : 1,
    "fail" : 0 })
print(df)
```

```python
x = df[["studyhours","attendance","internet"]]
y = df["result"]
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2, random_state = 42)
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
model = LogisticRegression()
model.fit(x_train_scaled,y_train)
y_pred = model.predict(x_test_scaled)
```

```python
comparsion = pd.DataFrame({"predcom" : y_test.values})
print(comparsion)
```

```python
y_probility = model.predict_proba(x_test_scaled)
print(y_probility)
```

```python
accuracy = accuracy_score(y_test,y_pred)
```

```python
cm = confusion_matrix(y_test,y_pred)
print(cm)
```

```python
print(classification_report(y_test, y_pred))
```
