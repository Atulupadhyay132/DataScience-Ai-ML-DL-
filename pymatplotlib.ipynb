```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "gender": [
        "Male", "Male", "Male", "Male",
        "Female", "Female", "Female", "Female"
    ],
    "marks": [
        50, 60, 70, 80,
        60, 75, 85, 95
    ]
})

sns.violinplot(
    x="gender",
    y="marks",
    data=df
)

plt.title("Marks Distribution by Gender")
plt.show()
```
